# Table 4: 80% threshold

One pooled OLS model with a common intercept and three category-specific forecast-revision slopes. Pairwise slope tests are native Stata `test` statements.

```stata
version 18
set more off
adopath ++ "../../../../reproductions/resources/ado"
global y "ret"
global x "fr_sup fr_norm fr_inf"
use "data/revisions/reg.dta", clear
isid event_id
assert !missing($y, fr_sup, fr_norm, fr_inf)
est clear
regress $y $x
quietly count if e(sample) & year(actdats) >= 1990 & year(actdats) <= 1994
local original_n = r(N)
quietly count if e(sample) & year(actdats) >= 1995
local added_n = r(N)
quietly summarize actdats if e(sample) & year(actdats) >= 1995, meanonly
local lastdate = r(max)
local last_added = string(`lastdate', "%tdCCYY-NN-DD")
display "EXT_SAMPLE s1 `original_n' `added_n' `last_added'"
est store table4_80
test fr_sup = fr_norm
test fr_sup = fr_inf
test fr_norm = fr_inf
esttab table4_80 using "tables/table4_80_threshold.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab table4_80 using "tables/table4_80_threshold.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab table4_80 using "tables/table4_80_threshold.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab table4_80 using "tables/table4_80_threshold.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
est clear
```
