# Table 2 Panel A: time trends
Seven OLS time-trend regressions on the common full-sample outcome file. The source does not report an estimator or standard-error correction; see `design.md`.

```stata
version 18
set more off
use "data/reg.dta", clear

est clear
label variable time "Time"
label variable sca "SCA"
label variable sox "SOX"
label variable abs_da "ABS DA"
label variable positive_da "Positive DA"
label variable negative_da "Negative DA"
label variable r_prod "R PROD"
label variable r_cfo "R CFO"
label variable r_disx "R DISX"
label variable rm_proxy "RM PROXY"

* Counts use the fit's active e(sample), at the firm-year level.
regress abs_da time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig1 = r(N)
quietly count if e(sample) & fyear >= 2006
local add1 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last1 = r(max)
display "EXT_SAMPLE s1 `orig1' `add1' `last1'"
estadd scalar original_n = `orig1'
estadd scalar added_n = `add1'
estadd scalar last_added = `last1'
est store m1

regress positive_da time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig2 = r(N)
quietly count if e(sample) & fyear >= 2006
local add2 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last2 = r(max)
display "EXT_SAMPLE s2 `orig2' `add2' `last2'"
estadd scalar original_n = `orig2'
estadd scalar added_n = `add2'
estadd scalar last_added = `last2'
est store m2

regress negative_da time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig3 = r(N)
quietly count if e(sample) & fyear >= 2006
local add3 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last3 = r(max)
display "EXT_SAMPLE s3 `orig3' `add3' `last3'"
estadd scalar original_n = `orig3'
estadd scalar added_n = `add3'
estadd scalar last_added = `last3'
est store m3

regress r_prod time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig4 = r(N)
quietly count if e(sample) & fyear >= 2006
local add4 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last4 = r(max)
display "EXT_SAMPLE s4 `orig4' `add4' `last4'"
estadd scalar original_n = `orig4'
estadd scalar added_n = `add4'
estadd scalar last_added = `last4'
est store m4

regress r_cfo time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig5 = r(N)
quietly count if e(sample) & fyear >= 2006
local add5 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last5 = r(max)
display "EXT_SAMPLE s5 `orig5' `add5' `last5'"
estadd scalar original_n = `orig5'
estadd scalar added_n = `add5'
estadd scalar last_added = `last5'
est store m5

regress r_disx time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig6 = r(N)
quietly count if e(sample) & fyear >= 2006
local add6 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last6 = r(max)
display "EXT_SAMPLE s6 `orig6' `add6' `last6'"
estadd scalar original_n = `orig6'
estadd scalar added_n = `add6'
estadd scalar last_added = `last6'
est store m6

regress rm_proxy time sca sox
quietly count if e(sample) & inrange(fyear, 1987, 2005)
local orig7 = r(N)
quietly count if e(sample) & fyear >= 2006
local add7 = r(N)
quietly summarize fyear if e(sample) & fyear >= 2006, meanonly
local last7 = r(max)
display "EXT_SAMPLE s7 `orig7' `add7' `last7'"
estadd scalar original_n = `orig7'
estadd scalar added_n = `add7'
estadd scalar last_added = `last7'
est store m7

esttab m1 m2 m3 m4 m5 m6 m7 using "tables/table2_panel_a.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N original_n added_n last_added r2 r2_a, fmt(%9.0f %9.0f %9.0f %9.0f %9.3f %9.3f) labels("Observations" "Original-period N" "Added-period N" "Last added FYEAR" "R-squared" "Adjusted R-squared")) replace label mtitle("ABS DA" "Positive DA" "Negative DA" "R PROD" "R CFO" "R DISX" "RM PROXY")
esttab m1 m2 m3 m4 m5 m6 m7 using "tables/table2_panel_a.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N original_n added_n last_added r2 r2_a, fmt(%9.0f %9.0f %9.0f %9.0f %9.3f %9.3f) labels("Observations" "Original-period N" "Added-period N" "Last added FYEAR" "R-squared" "Adjusted R-squared")) replace label mtitle("ABS DA" "Positive DA" "Negative DA" "R PROD" "R CFO" "R DISX" "RM PROXY")
esttab m1 m2 m3 m4 m5 m6 m7 using "tables/table2_panel_a.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N original_n added_n last_added r2 r2_a, fmt(%9.0f %9.0f %9.0f %9.0f %9.3f %9.3f) labels("Observations" "Original-period N" "Added-period N" "Last added FYEAR" "R-squared" "Adjusted R-squared")) replace label mtitle("ABS DA" "Positive DA" "Negative DA" "R PROD" "R CFO" "R DISX" "RM PROXY")
esttab m1 m2 m3 m4 m5 m6 m7 using "tables/table2_panel_a.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N original_n added_n last_added r2 r2_a, fmt(%9.0f %9.0f %9.0f %9.0f %9.3f %9.3f) labels("Observations" "Original-period N" "Added-period N" "Last added FYEAR" "R-squared" "Adjusted R-squared")) replace label mtitle("ABS DA" "Positive DA" "Negative DA" "R PROD" "R CFO" "R DISX" "RM PROXY")

est clear
```
