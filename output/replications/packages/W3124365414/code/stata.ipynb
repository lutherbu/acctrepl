# Roychowdhury (2006), Table 4

Three Fama-MacBeth regressions for abnormal CFO, discretionary expenses, and production costs.

```stata
version 18
set more off
adopath ++ "../../../../reproductions/resources/ado"
use "data/reg.dta", clear

egen long panel_id = group(gvkey)
xtset panel_id fyear

label variable ab_cfo "Abnormal CFO"
label variable ab_disexp "Abnormal discretionary expenses"
label variable ab_prod "Abnormal production costs"
label variable size "SIZE"
label variable mtb "MTB"
label variable netincome "Net income"
label variable suspect_ni "SUSPECT_NI"

est clear

xtfmb ab_cfo size mtb netincome suspect_ni, lag(1)
local total_s1 = e(N)
local periods_s1 = e(N_g)
tempvar es_s1
gen byte `es_s1' = e(sample)
quietly count if `es_s1' & fyear <= 2001
local original_s1 = r(N)
quietly count if `es_s1' & inrange(fyear,2002,2006)
local added_s1 = r(N)
quietly summarize fyear if `es_s1' & inrange(fyear,2002,2006), meanonly
local last_s1 = r(max)
display "EXT_SAMPLE s1 " `original_s1' " " `added_s1' " " %4.0f `last_s1'
estadd scalar original_n = `original_s1'
estadd scalar added_n = `added_s1'
estadd scalar last_added = `last_s1'
est store m_cfo
esttab m_cfo using "tables/extended_s1.csv", cells(b(star fmt(6)) p(par fmt(6))) stats(N N_g original_n added_n last_added, fmt(0 0 0 0 0) labels("Firm-year rows in e(sample)" "Annual coefficient fits" "Original firm-years" "Added firm-years" "Last added fiscal year")) replace label mtitles("Abnormal CFO")

xtfmb ab_disexp size mtb netincome suspect_ni, lag(1)
local total_s2 = e(N)
local periods_s2 = e(N_g)
tempvar es_s2
gen byte `es_s2' = e(sample)
quietly count if `es_s2' & fyear <= 2001
local original_s2 = r(N)
quietly count if `es_s2' & inrange(fyear,2002,2006)
local added_s2 = r(N)
quietly summarize fyear if `es_s2' & inrange(fyear,2002,2006), meanonly
local last_s2 = r(max)
display "EXT_SAMPLE s2 " `original_s2' " " `added_s2' " " %4.0f `last_s2'
estadd scalar original_n = `original_s2'
estadd scalar added_n = `added_s2'
estadd scalar last_added = `last_s2'
est store m_disexp
esttab m_disexp using "tables/extended_s2.csv", cells(b(star fmt(6)) p(par fmt(6))) stats(N N_g original_n added_n last_added, fmt(0 0 0 0 0) labels("Firm-year rows in e(sample)" "Annual coefficient fits" "Original firm-years" "Added firm-years" "Last added fiscal year")) replace label mtitles("Abnormal discretionary expenses")

xtfmb ab_prod size mtb netincome suspect_ni, lag(1)
local total_s3 = e(N)
local periods_s3 = e(N_g)
tempvar es_s3
gen byte `es_s3' = e(sample)
quietly count if `es_s3' & fyear <= 2001
local original_s3 = r(N)
quietly count if `es_s3' & inrange(fyear,2002,2006)
local added_s3 = r(N)
quietly summarize fyear if `es_s3' & inrange(fyear,2002,2006), meanonly
local last_s3 = r(max)
display "EXT_SAMPLE s3 " `original_s3' " " `added_s3' " " %4.0f `last_s3'
estadd scalar original_n = `original_s3'
estadd scalar added_n = `added_s3'
estadd scalar last_added = `last_s3'
est store m_prod
esttab m_prod using "tables/extended_s3.csv", cells(b(star fmt(6)) p(par fmt(6))) stats(N N_g original_n added_n last_added, fmt(0 0 0 0 0) labels("Firm-year rows in e(sample)" "Annual coefficient fits" "Original firm-years" "Added firm-years" "Last added fiscal year")) replace label mtitles("Abnormal production costs")

esttab m_cfo m_disexp m_prod using "tables/table4_main.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("Abnormal CFO" "Abnormal discretionary expenses" "Abnormal production costs") varlabels(_cons "Intercept" size "SIZE" mtb "MTB" netincome "Net income" suspect_ni "SUSPECT_NI") addnotes("Annual OLS cross-sections combined using Fama-MacBeth; Newey-West HAC lag 1." "N is the aggregate number of complete firm-years; annual counts are not reported in the source.")
esttab m_cfo m_disexp m_prod using "tables/table4_main.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("Abnormal CFO" "Abnormal discretionary expenses" "Abnormal production costs") varlabels(_cons "Intercept" size "SIZE" mtb "MTB" netincome "Net income" suspect_ni "SUSPECT_NI") addnotes("Annual OLS cross-sections combined using Fama-MacBeth; Newey-West HAC lag 1." "N is the aggregate number of complete firm-years; annual counts are not reported in the source.")
esttab m_cfo m_disexp m_prod using "tables/table4_main.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("Abnormal CFO" "Abnormal discretionary expenses" "Abnormal production costs") varlabels(_cons "Intercept" size "SIZE" mtb "MTB" netincome "Net income" suspect_ni "SUSPECT_NI") addnotes("Annual OLS cross-sections combined using Fama-MacBeth; Newey-West HAC lag 1." "N is the aggregate number of complete firm-years; annual counts are not reported in the source.")
esttab m_cfo m_disexp m_prod using "tables/table4_main.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("Abnormal CFO" "Abnormal discretionary expenses" "Abnormal production costs") varlabels(_cons "Intercept" size "SIZE" mtb "MTB" netincome "Net income" suspect_ni "SUSPECT_NI") addnotes("Annual OLS cross-sections combined using Fama-MacBeth; Newey-West HAC lag 1." "N is the aggregate number of complete firm-years; annual counts are not reported in the source.")
est clear
```
