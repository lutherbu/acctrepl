# Table 2 analysis
The year indicators use calendar year 2000 as the omitted category, matching the equation's reported 1988–1999 indicator family while retaining the 1988–2000 sample. The four signed truncated regressions run in independent cells so that each scope is attempted even if another likelihood fails.


```stata
version 18
set more off
adopath ++ "../../../../reproductions/resources/ado"
global ctrl "tenure age size industry_growth cash_flow auditor_type"
global fe "i.sic2 ib2000.calyear"
use "data/main/reg.dta", clear
summarize abs_da abs_ca, detail
est clear
regress abs_da $ctrl $fe
quietly count if e(sample) & inrange(fyear,1988,2000)
local s1_original = r(N)
quietly count if e(sample) & inrange(fyear,2001,2005)
local s1_added = r(N)
quietly summarize fyear if e(sample), meanonly
local s1_last = r(max)
display "EXT_SAMPLE s1 `s1_original' `s1_added' " %4.0f `s1_last'
est store m1
regress abs_ca $ctrl $fe
quietly count if e(sample) & inrange(fyear,1988,2000)
local s2_original = r(N)
quietly count if e(sample) & inrange(fyear,2001,2005)
local s2_added = r(N)
quietly summarize fyear if e(sample), meanonly
local s2_last = r(max)
display "EXT_SAMPLE s2 `s2_original' `s2_added' " %4.0f `s2_last'
est store m2
esttab m1 m2 using "tables/table2_panelA.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 m2 using "tables/table2_panelA.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 m2 using "tables/table2_panelA.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 m2 using "tables/table2_panelA.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
est clear
```

```stata
version 18
set more off
adopath ++ "../../../../reproductions/resources/ado"
global ctrl "tenure age size industry_growth cash_flow auditor_type"
global fe "i.sic2 ib2000.calyear"
use "data/main/reg.dta", clear
est clear
count if discretionary_accrual > 0
regress discretionary_accrual $ctrl $fe if discretionary_accrual > 0
predict double start_mean if e(sample), xb
summarize discretionary_accrual if e(sample), meanonly
local start_sigma = 2*abs(r(min))
if 2*abs(r(max)) > `start_sigma' local start_sigma = 2*abs(r(max))
summarize start_mean if e(sample), meanonly
if 2*abs(r(min)) > `start_sigma' local start_sigma = 2*abs(r(min))
if 2*abs(r(max)) > `start_sigma' local start_sigma = 2*abs(r(max))
matrix b0 = e(b)
matrix b0 = b0, (`start_sigma')
truncreg discretionary_accrual $ctrl $fe if discretionary_accrual > 0, ll(0) technique(bhhh 20 nr 300) vce(oim) from(b0, copy)
quietly count if e(sample) & inrange(fyear,1988,2000)
local s3_original = r(N)
quietly count if e(sample) & inrange(fyear,2001,2005)
local s3_added = r(N)
quietly summarize fyear if e(sample), meanonly
local s3_last = r(max)
display "EXT_SAMPLE s3 `s3_original' `s3_added' " %4.0f `s3_last'
est store m1
esttab m1 using "tables/table2_panelB_posDA.md", mtitle("Positive Discretionary Accruals") cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 using "tables/table2_panelB_posDA.rtf", mtitle("Positive Discretionary Accruals") cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 using "tables/table2_panelB_posDA.csv", mtitle("Positive Discretionary Accruals") cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
esttab m1 using "tables/table2_panelB_posDA.tex", mtitle("Positive Discretionary Accruals") cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label drop(*.sic2 *.calyear)
est clear
```
