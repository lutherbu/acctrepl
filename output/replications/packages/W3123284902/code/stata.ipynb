# Stata Fama–MacBeth estimates

Each cell loads the assigned firm-month handoff independently. Monthly cross-sections use unweighted OLS with a constant; `xtfmb` averages monthly slopes and uses the time-series covariance across fitted months. SAS logs document monthly complete-case counts and full-rank checks.

```stata
version 18
set more off
adopath ++ "../../../../reproductions/resources/ado"

global y "ret"
global x "lnsize lnbm retm1 ret12_2 ret36_13"
global acc "accr1"
global noa "noa1"
global elig "fit_a"

use "data/noa_firm_month/reg.dta", clear
format mdate %tm
tsset firmid mdate

est clear
xtfmb $y $x $acc $noa if $elig==1

tempvar used_s1
generate byte `used_s1' = e(sample)
quietly count if `used_s1'==1 & mdate < tm(2003m1)
local original_n = r(N)
quietly count if `used_s1'==1 & mdate >= tm(2003m1)
local added_n = r(N)
local last_period "none"
if `added_n' > 0 {
    quietly summarize mdate if `used_s1'==1 & mdate >= tm(2003m1), meanonly
    local last_m = r(max)
    local last_year = year(dofm(`last_m'))
    local last_month = month(dofm(`last_m'))
    local last_period = string(`last_year',"%04.0f") + "-" + string(`last_month',"%02.0f")
}
display "EXT_SAMPLE s1 `original_n' `added_n' `last_period'"
local fit_n = e(N)
local fmb_months = e(N_g)
local b_noa : display %21.12g _b[noa1]
local b_accr : display %21.12g _b[accr1]
estadd scalar fmb_months = e(N_g)
estadd scalar original_n = `original_n'
estadd scalar added_n = `added_n'
est store m_a
esttab m_a using "artifacts/s1_aggregate.csv", cells(b(fmt(6))) stats(N fmb_months original_n added_n r2, fmt(0 0 0 0 6) labels("Firm-month observations" "FMB months" "Original-period firm-months" "Added-period firm-months" "Average monthly R-squared")) replace label
esttab m_a using "tables/table5_panel_a_model4.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(N fmb_months r2, fmt(0 0 3) labels("Firm-month observations" "FMB months" "Average monthly R-squared")) r2 ar2 replace label mtitles("Panel A: 1-year lag")
esttab m_a using "tables/table5_panel_a_model4.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(N fmb_months r2, fmt(0 0 3) labels("Firm-month observations" "FMB months" "Average monthly R-squared")) r2 ar2 replace label mtitles("Panel A: 1-year lag")
esttab m_a using "tables/table5_panel_a_model4.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(N fmb_months r2, fmt(0 0 3) labels("Firm-month observations" "FMB months" "Average monthly R-squared")) r2 ar2 replace label mtitles("Panel A: 1-year lag")
esttab m_a using "tables/table5_panel_a_model4.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(N fmb_months r2, fmt(0 0 3) labels("Firm-month observations" "FMB months" "Average monthly R-squared")) r2 ar2 replace label mtitles("Panel A: 1-year lag")
est clear
```

## Descriptive statistics by panel sample

These tables report N, mean, standard deviation, and quartiles for each model’s complete-case sample. Each descriptive cell uses the corresponding sample flag from SAS.

```stata
version 18
set more off

global ctrl "ret lnsize lnbm retm1 ret12_2 ret36_13"

use "data/noa_firm_month/reg.dta", clear
format mdate %tm

est clear
global acc "accr1"
global noa "noa1"
estpost tabstat $ctrl $acc $noa if fit_a==1, stat(N mean sd p25 p50 p75) c(s)
esttab using "tables/descriptive_panel_a.md", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_a.rtf", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_a.csv", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_a.tex", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
est clear

global acc "accr2"
global noa "noa2"
estpost tabstat $ctrl $acc $noa if fit_b==1, stat(N mean sd p25 p50 p75) c(s)
esttab using "tables/descriptive_panel_b.md", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_b.rtf", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_b.csv", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_b.tex", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
est clear

global acc "accr3"
global noa "noa3"
estpost tabstat $ctrl $acc $noa if fit_c==1, stat(N mean sd p25 p50 p75) c(s)
esttab using "tables/descriptive_panel_c.md", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_c.rtf", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_c.csv", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/descriptive_panel_c.tex", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
est clear
```
