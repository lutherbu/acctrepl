# Stata analysis: Table 3, pooled 1993–2003 levels

Each model is loaded and estimated in an independent cell. The cross-sectional estimator and covariates are retained; exported p-values are two-sided, while the paper reports one-sided p-values where predicted.

```stata
version 18
set more off

global y "conmkt"
global x "stdroa div lev roa size salesgro rndadv"
global sample "elig_mkt"

use "data/main/reg.dta", clear
est clear
regress $y $x if $sample == 1, vce(robust)
gen byte __period_original = (n_conmkt_original>0 | n_controls_original>0 | n_market_original>0)
gen byte __period_added = (n_conmkt_added>0 | n_controls_added>0 | n_market_added>0)
quietly count if e(sample) & __period_original
local original_n = r(N)
quietly count if e(sample) & __period_added
local added_n = r(N)
egen __last_added = rowmax(last_conmkt_added last_controls_added last_market_added)
quietly summarize __last_added if e(sample) & __period_added, meanonly
local last_added = floor(r(max))
display as text "EXT_SAMPLE s1 `original_n' `added_n' `last_added'"
drop __period_original __period_added __last_added
est store panel_a

esttab panel_a using "tables/table3_panel_a_levels_1993_98.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table3_panel_a_levels_1993_98.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table3_panel_a_levels_1993_98.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table3_panel_a_levels_1993_98.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
est clear
```
