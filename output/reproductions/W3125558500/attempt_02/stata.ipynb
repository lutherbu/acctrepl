# Stata Analysis Code Template

This is a template for Stata analysis code. It is designed to be used in a Jupyter notebook.

* Key rules: 
    - EVERY cell should be self-contained and executable, so all global settings should be repeated in every cell.
    - Each cell usually write one table, panel or figure.
    - Do not write custom `program define` wrappers, Mata code, or meta/diagnostic machinery (postfiles, model-metadata logging, coefficient dumps); call estimators directly and store results with `est store` (plus `estadd` for table rows).
    - Keep the codes as few lines as possible, minimize unnecessary code beyond shown in the examples below.

* Output rules (always follow):
    - Tables: export to four formats with the same basename — `.md`, `.rtf`, `.csv`, `.tex` — using the `esttab` options shown below (`cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label`; add `r2_p` for nonlinear models).
    - Figures: export to both `.pdf` and `.png`.
    - No extra side files (`.ster`, metadata, diagnostic dumps); the four formats are the deliverable.

* Entry file: one `reg.dta` file per sample (one for a project that uses only one baseline sample)

## Table 4, Panel A: NI on PRETAX regions

```stata
version 18
set more off

global dep "ni_usd"
global x "pretax_l pretax_sl pretax_sp pretax_p"

use "data/reg.dta", clear
label var ni_usd "Net income (USD millions)"
label var pretax_l "Pretax income in loss region (USD millions)"
label var pretax_sl "Pretax income in small-loss region (USD millions)"
label var pretax_sp "Pretax income in small-profit region (USD millions)"
label var pretax_p "Pretax income in profit region (USD millions)"

est clear
regress $dep $x if sample_common == 1
est store panel_a

test pretax_sl = pretax_sp
estadd scalar F_SLSP = r(F)
estadd scalar P_SLSP = r(p)
test pretax_l = pretax_p
estadd scalar F_LP = r(F)
estadd scalar P_LP = r(p)

esttab panel_a using "tables/table4_panel_a.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a F_SLSP P_SLSP F_LP P_LP, labels("Observations" "R-squared" "Adjusted R-squared" "F: beta_SL = beta_SP" "p: beta_SL = beta_SP" "F: beta_L = beta_P" "p: beta_L = beta_P") fmt(0 3 3 2 3 2 3)) replace label
esttab panel_a using "tables/table4_panel_a.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a F_SLSP P_SLSP F_LP P_LP, labels("Observations" "R-squared" "Adjusted R-squared" "F: beta_SL = beta_SP" "p: beta_SL = beta_SP" "F: beta_L = beta_P" "p: beta_L = beta_P") fmt(0 3 3 2 3 2 3)) replace label
esttab panel_a using "tables/table4_panel_a.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a F_SLSP P_SLSP F_LP P_LP, labels("Observations" "R-squared" "Adjusted R-squared" "F: beta_SL = beta_SP" "p: beta_SL = beta_SP" "F: beta_L = beta_P" "p: beta_L = beta_P") fmt(0 3 3 2 3 2 3)) replace label
esttab panel_a using "tables/table4_panel_a.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a F_SLSP P_SLSP F_LP P_LP, labels("Observations" "R-squared" "Adjusted R-squared" "F: beta_SL = beta_SP" "p: beta_SL = beta_SP" "F: beta_L = beta_P" "p: beta_L = beta_P") fmt(0 3 3 2 3 2 3)) replace label
est clear
```

## Panel A: descriptive statistics for the estimation sample

```stata
version 18
set more off

global desc "ni_usd pretax_l pretax_sl pretax_sp pretax_p"

use "data/reg.dta", clear
label var ni_usd "Net income (USD millions)"
label var pretax_l "Pretax income in loss region (USD millions)"
label var pretax_sl "Pretax income in small-loss region (USD millions)"
label var pretax_sp "Pretax income in small-profit region (USD millions)"
label var pretax_p "Pretax income in profit region (USD millions)"

est clear
estpost tabstat $desc if sample_common == 1, stat(N mean sd p25 p50 p75) c(s)
esttab using "tables/table4_panel_a_descriptive.md", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/table4_panel_a_descriptive.rtf", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/table4_panel_a_descriptive.csv", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
esttab using "tables/table4_panel_a_descriptive.tex", replace cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f))") label
est clear
```
