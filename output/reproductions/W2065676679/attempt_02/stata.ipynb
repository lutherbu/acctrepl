# Beaver and Ryan (2000), Table 5 and descriptive comparisons

Every Table 5 horizon uses the same declared complete-five-future-ROE sample. Panel B runs the native robust Wald tests only after SAS verifies the component sum identity.

```stata
version 18
set more off
global future "roe1 roe2 roe3 roe4 roe5"
use "data/main/reg.dta", clear
est clear
local h = 1
foreach y of varlist $future {
    regress `y' roe0 btm, vce(robust)
    est store a`h'
    local ++h
}
esttab a1 a2 a3 a4 a5 using "tables/table5_panel_a.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab a1 a2 a3 a4 a5 using "tables/table5_panel_a.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab a1 a2 a3 a4 a5 using "tables/table5_panel_a.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab a1 a2 a3 a4 a5 using "tables/table5_panel_a.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
est clear
```

```stata
version 18
set more off
global future "roe1 roe2 roe3 roe4 roe5"
use "data/main/reg.dta", clear
est clear
summarize identity_pass, meanonly
local identityok = (r(min) == 1 & r(max) == 1)
local h = 1
foreach y of varlist $future {
    regress `y' roe0 bc lc timefx eps, vce(robust)
    lincom bc - lc
    local cdiff = r(estimate)
    local cdt = r(t)
    local cdp = r(p)
    estadd scalar cdiff = `cdiff'
    estadd scalar cdt = `cdt'
    estadd scalar cdp = `cdp'
    if `identityok' {
        test (bc = lc) (bc = timefx) (bc = eps)
        local incrF = r(F)
        local incrp = r(p)
        estadd scalar incrF = `incrF'
        estadd scalar incrp = `incrp'
    }
    else {
        estadd scalar incrF = .
        estadd scalar incrp = .
    }
    est store b`h'
    local ++h
}
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(cdiff cdt incrF N, fmt(3 2 2 0) labels("c-d" "t(c-d)" "Incremental robust F" "N")) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(cdiff cdt incrF N, fmt(3 2 2 0) labels("c-d" "t(c-d)" "Incremental robust F" "N")) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(cdiff cdt incrF N, fmt(3 2 2 0) labels("c-d" "t(c-d)" "Incremental robust F" "N")) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(cdiff cdt incrF N, fmt(3 2 2 0) labels("c-d" "t(c-d)" "Incremental robust F" "N")) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b_tests.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(cdiff cdt cdp incrF incrp N r2 ar2, fmt(3 2 3 2 3 0 3 3) labels("BC-LC" "t(BC-LC)" "p(BC-LC)" "Incremental robust F(3,df)" "p(incremental F)" "N" "R-squared" "Adjusted R-squared")) drop(*) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b_tests.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(cdiff cdt cdp incrF incrp N r2 ar2, fmt(3 2 3 2 3 0 3 3) labels("BC-LC" "t(BC-LC)" "p(BC-LC)" "Incremental robust F(3,df)" "p(incremental F)" "N" "R-squared" "Adjusted R-squared")) drop(*) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b_tests.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(cdiff cdt cdp incrF incrp N r2 ar2, fmt(3 2 3 2 3 0 3 3) labels("BC-LC" "t(BC-LC)" "p(BC-LC)" "Incremental robust F(3,df)" "p(incremental F)" "N" "R-squared" "Adjusted R-squared")) drop(*) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
esttab b1 b2 b3 b4 b5 using "tables/table5_panel_b_tests.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) stats(cdiff cdt cdp incrF incrp N r2 ar2, fmt(3 2 3 2 3 0 3 3) labels("BC-LC" "t(BC-LC)" "p(BC-LC)" "Incremental robust F(3,df)" "p(incremental F)" "N" "R-squared" "Adjusted R-squared")) drop(*) replace label mtitles("1 year" "2 years" "3 years" "4 years" "5 years")
est clear
```

```stata
version 18
set more off
global samplevars "roe0 btm roe1 roe2 roe3 roe4 roe5 bc lc timefx eps"
use "data/main/reg.dta", clear
est clear
estpost tabstat $samplevars, by(fyear) stat(N p50) c(s)
esttab using "tables/table5_sample_desc.md", replace cells("count(fmt(0)) p50(fmt(3))") label
esttab using "tables/table5_sample_desc.rtf", replace cells("count(fmt(0)) p50(fmt(3))") label
esttab using "tables/table5_sample_desc.csv", replace cells("count(fmt(0)) p50(fmt(3))") label
esttab using "tables/table5_sample_desc.tex", replace cells("count(fmt(0)) p50(fmt(3))") label
est clear
```

```stata
version 18
set more off
use "data/history/reg.dta", clear
encode gvkey, gen(firm_id)
xtset firm_id fyear
gen r0 = ret
forvalues j = 1/6 {
    gen r`j' = L`j'.ret
}
est clear
local h = 1
forvalues formation = 1984/1988 {
    preserve
    keep if inrange(fyear,1981,`formation')
    keep if !missing(btm,r0,r1,r2,r3,r4,r5,r6)
    bysort firm_id: gen nfit = _N
    keep if nfit >= 4
    reghdfe btm r0-r6, absorb(firm_id fyear)
    est store q`h'
    local ++h
    restore
}
esttab q1 q2 q3 q4 q5 using "tables/eq4_rank_audit.md", cells(b(fmt(3))) drop(*) stats(N df_m df_r, fmt(0 0 0) labels("Observations" "Model df" "Residual df")) replace label mtitles("1984" "1985" "1986" "1987" "1988")
esttab q1 q2 q3 q4 q5 using "tables/eq4_rank_audit.rtf", cells(b(fmt(3))) drop(*) stats(N df_m df_r, fmt(0 0 0) labels("Observations" "Model df" "Residual df")) replace label mtitles("1984" "1985" "1986" "1987" "1988")
esttab q1 q2 q3 q4 q5 using "tables/eq4_rank_audit.csv", cells(b(fmt(3))) drop(*) stats(N df_m df_r, fmt(0 0 0) labels("Observations" "Model df" "Residual df")) replace label mtitles("1984" "1985" "1986" "1987" "1988")
esttab q1 q2 q3 q4 q5 using "tables/eq4_rank_audit.tex", cells(b(fmt(3))) drop(*) stats(N df_m df_r, fmt(0 0 0) labels("Observations" "Model df" "Residual df")) replace label mtitles("1984" "1985" "1986" "1987" "1988")
est clear
```
