Table 3, columns (1) and (2): the pooled firm-year Eq. (7) models, estimated separately for DD and ABACC² proxies. The paper calls for GMM with three-lag Newey–West HAC inference.

Because the paper leaves instruments unspecified, the best-effort linear moment set uses the included RHS regressors as instruments. This is exactly identified, so its coefficient solution equals OLS. Stata’s installed `ivreg2` supports panel HAC after `xtset`; `robust bw(4) kernel(bartlett)` applies the Bartlett kernel with three nonzero lags. A1’s `ivregress gmm` panel attempt returned r(459), and direct `gmm` did not converge with usable HAC standard errors. A2’s direct `ivreg2` fits completed for both scopes; inference and instrument assumptions are documented in `design.md` and `review.md`.
