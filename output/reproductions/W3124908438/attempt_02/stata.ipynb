# Stata estimation

Each fit uses the common SAS-built firm-year sample and annual Fama–MacBeth cross-sections with one Newey–West lag, matching the paper's annual coefficient averaging. No pooled regression, fixed effects, or weights are added. `xtfmb` reports mean raw R-squared; the paper's adjusted R-squared is unavailable from the installed estimator. The code uses the relative path to the installed `xtfmb` package and loads the common `reg.dta` in every code cell.

```stata
version 18
set more off
use "data/reg.dta", clear
est clear
estpost summarize rt ur rfut earn dearn rnoa drnoa pm ato dpm dato rnoap1 pmp1 atop1 sur anrev, detail
esttab using "tables/descriptives.md", cells("count(fmt(0)) mean(fmt(3)) sd(fmt(3)) p50(fmt(3)) min(fmt(3)) max(fmt(3))") replace label
esttab using "tables/descriptives.rtf", cells("count(fmt(0)) mean(fmt(3)) sd(fmt(3)) p50(fmt(3)) min(fmt(3)) max(fmt(3))") replace label
esttab using "tables/descriptives.csv", cells("count(fmt(0)) mean(fmt(3)) sd(fmt(3)) p50(fmt(3)) min(fmt(3)) max(fmt(3))") replace label
esttab using "tables/descriptives.tex", cells("count(fmt(0)) mean(fmt(3)) sd(fmt(3)) p50(fmt(3)) min(fmt(3)) max(fmt(3))") replace label
est clear
```

```stata
version 18
set more off
adopath ++ "../../ado"
use "data/reg.dta", clear
egen long firmid = group(gvkey)
xtset firmid fyear
est clear
xtfmb rt earn dearn rnoa drnoa pm ato dpm dato rnoap1 pmp1 atop1, lag(1)
est store m1
esttab m1 using "tables/table5_full_future_components.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 replace label
esttab m1 using "tables/table5_full_future_components.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 replace label
esttab m1 using "tables/table5_full_future_components.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 replace label
esttab m1 using "tables/table5_full_future_components.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 replace label
est clear
```
