# Earnings volatility and earnings predictability

The three regression tables report all distinct baseline fits from a random firm-level exploratory split. The h=1 full-sample and extreme-quintile fits are shared by Tables 2 and 3. Contrast tests are supporting inference; the OLS fit rows remain separate by horizon and population.


## Table 2: one-year persistence


```stata
version 18
set more off
adopath ++ "../../ado"
use "data/reg.dta", clear
est clear
reg future_earn1 earn_t
est store full_h1
reg future_earn1 earn_t if vol_q==1
est store q1_h1
reg future_earn1 earn_t if vol_q==2
est store q2_h1
reg future_earn1 earn_t if vol_q==3
est store q3_h1
reg future_earn1 earn_t if vol_q==4
est store q4_h1
reg future_earn1 earn_t if vol_q==5
est store q5_h1
esttab full_h1 q1_h1 q2_h1 q3_h1 q4_h1 q5_h1 using "tables/persistence_h1.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a, fmt(0 3 3) labels("N" "R-squared" "Adjusted R-squared")) replace label drop(_cons) mtitles("Full h=1" "Q1 h=1" "Q2 h=1" "Q3 h=1" "Q4 h=1" "Q5 h=1")
esttab full_h1 q1_h1 q2_h1 q3_h1 q4_h1 q5_h1 using "tables/persistence_h1.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a, fmt(0 3 3) labels("N" "R-squared" "Adjusted R-squared")) replace label drop(_cons) mtitles("Full h=1" "Q1 h=1" "Q2 h=1" "Q3 h=1" "Q4 h=1" "Q5 h=1")
esttab full_h1 q1_h1 q2_h1 q3_h1 q4_h1 q5_h1 using "tables/persistence_h1.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a, fmt(0 3 3) labels("N" "R-squared" "Adjusted R-squared")) replace label drop(_cons) mtitles("Full h=1" "Q1 h=1" "Q2 h=1" "Q3 h=1" "Q4 h=1" "Q5 h=1")
esttab full_h1 q1_h1 q2_h1 q3_h1 q4_h1 q5_h1 using "tables/persistence_h1.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a, fmt(0 3 3) labels("N" "R-squared" "Adjusted R-squared")) replace label drop(_cons) mtitles("Full h=1" "Q1 h=1" "Q2 h=1" "Q3 h=1" "Q4 h=1" "Q5 h=1")
est clear
```

## Supporting extreme-quintile contrast tests

The interaction tests use Q1 minus Q5 coefficient differences; their native t-test p-values are two-sided.


```stata
version 18
set more off
adopath ++ "../../ado"
use "data/reg.dta", clear
est clear
reg future_earn1 c.earn_t##ib1.vol_q if inlist(vol_q,1,5)
test 5.vol_q#c.earn_t = 0
local t2_p = r(p)
local t2_diff = -_b[5.vol_q#c.earn_t]
estadd scalar beta_diff = `t2_diff'
estadd scalar p_diff = `t2_p'
est store diff_t2
reg sum_earn5 c.earn_t##ib1.vol_q if inlist(vol_q,1,5)
test 5.vol_q#c.earn_t = 0
local t4_p = r(p)
local t4_diff = -_b[5.vol_q#c.earn_t]
estadd scalar beta_diff = `t4_diff'
estadd scalar p_diff = `t4_p'
est store diff_t4
esttab diff_t2 diff_t4 using "tables/persistence_contrasts.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a beta_diff p_diff, fmt(0 3 3 3 3) labels("N" "R-squared" "Adjusted R-squared" "Q1 minus Q5 beta" "Two-sided p-value")) replace label mtitles("Table 2 h=1" "Table 4 five-year sum")
esttab diff_t2 diff_t4 using "tables/persistence_contrasts.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a beta_diff p_diff, fmt(0 3 3 3 3) labels("N" "R-squared" "Adjusted R-squared" "Q1 minus Q5 beta" "Two-sided p-value")) replace label mtitles("Table 2 h=1" "Table 4 five-year sum")
esttab diff_t2 diff_t4 using "tables/persistence_contrasts.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a beta_diff p_diff, fmt(0 3 3 3 3) labels("N" "R-squared" "Adjusted R-squared" "Q1 minus Q5 beta" "Two-sided p-value")) replace label mtitles("Table 2 h=1" "Table 4 five-year sum")
esttab diff_t2 diff_t4 using "tables/persistence_contrasts.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 stats(N r2 r2_a beta_diff p_diff, fmt(0 3 3 3 3) labels("N" "R-squared" "Adjusted R-squared" "Q1 minus Q5 beta" "Two-sided p-value")) replace label mtitles("Table 2 h=1" "Table 4 five-year sum")
est clear
```
