# Native Stata analysis

Table 1 and actual Table 2 sample descriptions, then the two retained Table 2 column 6 regressions.

```stata
version 18
set more off

global t1vars "earnings_t opacc_t tacc_t spi_ratio cfo_t cfi_t cff_t fcf_t"
use "data/accrual_persistence_main/reg.dta", clear
keep if table1_sample==1

label var earnings_t "Earnings"
label var opacc_t "Operating accruals"
label var tacc_t "Total accruals"
label var spi_ratio "Special items"
label var cfo_t "CFO"
label var cfi_t "CFI"
label var cff_t "CFF"
label var fcf_t "FCF"

est clear
estpost tabstat $t1vars, stat(N mean sd p25 p50 p75 min max) c(s)
esttab using "tables/table1_actual_sample.md", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table1_actual_sample.rtf", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table1_actual_sample.csv", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table1_actual_sample.tex", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
est clear
correlate $t1vars
spearman $t1vars
```

```stata
version 18
set more off

global t2vars "earnings_t earnings_next cfo_t opacc_t fcf_t tacc_t spi_ratio si low_op low_tacc earn_low_op earn_low_op_si earn_low_tacc earn_low_tacc_si"
use "data/accrual_persistence_main/reg.dta", clear
keep if table2_sample==1

label var earnings_t "Earnings t"
label var earnings_next "Earnings t+1"
label var cfo_t "CFO t"
label var opacc_t "OPACC t"
label var fcf_t "FCF t"
label var tacc_t "TACC t"
label var spi_ratio "Special items"
label var si "SI"
label var low_op "Low OPACC"
label var low_tacc "Low TACC"
label var earn_low_op "Earnings x Low OPACC"
label var earn_low_op_si "Earnings x Low OPACC x SI"
label var earn_low_tacc "Earnings x Low TACC"
label var earn_low_tacc_si "Earnings x Low TACC x SI"

est clear
estpost tabstat $t2vars, stat(N mean sd p25 p50 p75 min max) c(s)
esttab using "tables/table2_actual_sample.md", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table2_actual_sample.rtf", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table2_actual_sample.csv", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
esttab using "tables/table2_actual_sample.tex", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f)) p25(fmt(%9.3f)) p50(fmt(%9.3f)) p75(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") label replace
est clear
```

```stata
version 18
set more off

global y "earnings_next"
global x_a "cfo_t opacc_t earn_low_op earn_low_op_si"
use "data/accrual_persistence_main/reg.dta", clear
keep if table2_sample==1

label var earnings_next "Earnings t+1"
label var cfo_t "Cash flows from operations"
label var opacc_t "Operating accruals"
label var earn_low_op "Earnings x Low OPACC"
label var earn_low_op_si "Earnings x Low OPACC x SI"

est clear
regress $y $x_a
est store panel_a
esttab panel_a using "tables/table2_panel_a.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table2_panel_a.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table2_panel_a.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_a using "tables/table2_panel_a.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
est clear
```

```stata
version 18
set more off

global y "earnings_next"
global x_b "fcf_t tacc_t earn_low_tacc earn_low_tacc_si"
use "data/accrual_persistence_main/reg.dta", clear
keep if table2_sample==1

label var earnings_next "Earnings t+1"
label var fcf_t "Free cash flow"
label var tacc_t "Total accruals"
label var earn_low_tacc "Earnings x Low TACC"
label var earn_low_tacc_si "Earnings x Low TACC x SI"

est clear
regress $y $x_b
est store panel_b
esttab panel_b using "tables/table2_panel_b.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_b using "tables/table2_panel_b.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_b using "tables/table2_panel_b.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab panel_b using "tables/table2_panel_b.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
est clear
```
