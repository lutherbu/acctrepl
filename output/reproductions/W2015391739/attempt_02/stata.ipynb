# Stata baseline outputs

Two self-contained cells estimate the retained Table 2 Panel A and Panel B fits. The other cells export full-sample and sign-partition descriptives for the actual sample inputs.

```stata
version 18
set more off
use "data/basu_replication_1963_1990/reg.dta", clear
label variable x "Scaled earnings X"
label variable r_actual "Actual return"
label variable r_sim "Simulated return"
est clear
reg x r_actual if r_actual < 0
est store bad_actual
reg x r_sim if !missing(r_sim) & r_sim < 0
est store bad_sim
esttab bad_actual bad_sim using "tables/table2_panelA.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab bad_actual bad_sim using "tables/table2_panelA.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab bad_actual bad_sim using "tables/table2_panelA.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab bad_actual bad_sim using "tables/table2_panelA.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
est clear
```

```stata
version 18
set more off
use "data/basu_replication_1963_1990/reg.dta", clear
label variable x "Scaled earnings X"
label variable r_actual "Actual return"
label variable r_sim "Simulated return"
est clear
reg x r_actual if !missing(r_actual) & r_actual >= 0
est store good_actual
reg x r_sim if !missing(r_sim) & r_sim >= 0
est store good_sim
esttab good_actual good_sim using "tables/table2_panelB.md", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab good_actual good_sim using "tables/table2_panelB.rtf", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab good_actual good_sim using "tables/table2_panelB.csv", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
esttab good_actual good_sim using "tables/table2_panelB.tex", cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 drop(_cons) replace label mtitle("Actual returns" "Simulated returns")
est clear
```

```stata
version 18
set more off
use "data/basu_replication_1963_1990/reg.dta", clear
label variable x "X: scaled earnings"
label variable r_actual "R: actual return"
label variable r_sim "R: simulated return"
est clear
estpost tabstat x r_actual r_sim, stat(N mean p50 variance min max) c(s)
esttab using "tables/table1_full_sample_descriptives.md", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) p50(fmt(%9.3f)) variance(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") replace label
esttab using "tables/table1_full_sample_descriptives.rtf", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) p50(fmt(%9.3f)) variance(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") replace label
esttab using "tables/table1_full_sample_descriptives.csv", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) p50(fmt(%9.3f)) variance(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") replace label
esttab using "tables/table1_full_sample_descriptives.tex", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) p50(fmt(%9.3f)) variance(fmt(%9.3f)) min(fmt(%9.3f)) max(fmt(%9.3f))") replace label
est clear
```

```stata
version 18
set more off
use "data/basu_replication_1963_1990/reg.dta", clear
label variable x "X: scaled earnings"
label variable r_actual "R: actual return"
label variable r_sim "R: simulated return"
est clear
estpost tabstat x r_actual r_sim if r_actual < 0, stat(N mean sd) c(s)
est store actual_bad
estpost tabstat x r_actual r_sim if !missing(r_sim) & r_sim < 0, stat(N mean sd) c(s)
est store simulated_bad
estpost tabstat x r_actual r_sim if !missing(r_actual) & r_actual >= 0, stat(N mean sd) c(s)
est store actual_good
estpost tabstat x r_actual r_sim if !missing(r_sim) & r_sim >= 0, stat(N mean sd) c(s)
est store simulated_good
esttab actual_bad simulated_bad actual_good simulated_good using "tables/table2_partition_descriptives.md", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f))") replace label mtitle("Actual bad" "Simulated bad" "Actual good" "Simulated good")
esttab actual_bad simulated_bad actual_good simulated_good using "tables/table2_partition_descriptives.rtf", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f))") replace label mtitle("Actual bad" "Simulated bad" "Actual good" "Simulated good")
esttab actual_bad simulated_bad actual_good simulated_good using "tables/table2_partition_descriptives.csv", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f))") replace label mtitle("Actual bad" "Simulated bad" "Actual good" "Simulated good")
esttab actual_bad simulated_bad actual_good simulated_good using "tables/table2_partition_descriptives.tex", cells("count(fmt(%9.0f)) mean(fmt(%9.3f)) sd(fmt(%9.3f))") replace label mtitle("Actual bad" "Simulated bad" "Actual good" "Simulated good")
est clear
```
