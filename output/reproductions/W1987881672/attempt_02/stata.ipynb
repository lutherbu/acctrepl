# Stata analysis — Table 1 fiscal-year trends

Each fit is a univariate OLS regression of the annual outcome on the FPEDATS fiscal-year proxy. The paper does not state the estimator or covariance estimator; conventional OLS inference is the documented best-effort choice.

```stata
version 18
set more off
adopath ++ "../../ado"
global y "med_all"
global x "fiscal_year"
use "data/ibes_surprises/reg.dta", clear
label variable $x "Fiscal year trend"
est clear
regress $y $x
est store m1
esttab m1 using "tables/median_surprise_earnings.md", keep(fiscal_year) cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab m1 using "tables/median_surprise_earnings.rtf", keep(fiscal_year) cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab m1 using "tables/median_surprise_earnings.csv", keep(fiscal_year) cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
esttab m1 using "tables/median_surprise_earnings.tex", keep(fiscal_year) cells(b(star fmt(3)) p(par)) starlevels(* 0.1 ** 0.05 *** 0.01) r2 ar2 replace label
est clear
```
